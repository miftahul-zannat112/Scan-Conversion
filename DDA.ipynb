```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
start_x = float(input("Enter x1: "))
start_y = float(input("Enter y1: "))

finish_x = float(input("Enter x2: "))
finish_y = float(input("Enter y2: "))

gradient = (finish_y - start_y) / (finish_x - start_x)

x_difference = finish_x - start_x
y_difference = finish_y - start_y

print("\nSlope =", round(gradient, 2))
print("Delta X =", x_difference)
print("Delta Y =", y_difference)

if abs(x_difference) >= abs(y_difference):

    x_increment = 1 if x_difference > 0 else -1
    y_increment = gradient * x_increment

else:

    y_increment = 1 if y_difference > 0 else -1
    x_increment = y_increment / gradient

print("Assigned X increment =", round(x_increment, 2))
print("Assigned Y increment =", round(y_increment, 2))

total_steps = int(max(abs(x_difference), abs(y_difference)))

current_x = start_x
current_y = start_y

dda_rows = []
```

```python
for counter in range(total_steps + 1):
    next_x = current_x + x_increment
    next_y = current_y + y_increment

    if counter < total_steps:
        dda_rows.append([
            round(current_x, 2),
            round(current_y, 2),
            round(next_x, 2),
            round(next_y, 2)
        ])

    current_x = next_x
    current_y = next_y

dda_table = pd.DataFrame(dda_rows, columns=[
    "X(i)",
    "Y(i)",
    "X(i+1)",
    "Y(i+1)"
])

print("\nDDA Calculation Table:")
display(dda_table)

print("\nConverted Points:")

current_x = start_x
current_y = start_y
pixel_points = []

for counter in range(total_steps + 1):
    pixel_points.append((round(current_x), round(current_y)))
    current_x += x_increment
    current_y += y_increment

print(pixel_points)

display_x = [item[0] for item in pixel_points]
display_y = [item[1] for item in pixel_points]

plt.figure(figsize=(7, 5))
plt.plot(display_x, display_y, marker='o', linewidth=2)

for px, py in pixel_points:
    plt.text(px + 0.05, py + 0.05, f"({px},{py})")

plt.grid(True)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("DDA Line Rasterization")

plt.show()
```
